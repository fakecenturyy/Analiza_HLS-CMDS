```python
import pandas as pd
import numpy as np
import librosa
import os
from scipy.signal import butter, lfilter

# Parametry Wejściowe
BASE_PATH = 'C:/Users/macie/Desktop/PROJEKT/'
PLIKI_CSV = ['train.csv', 'val.csv', 'test.csv']
# Filtracja pasmowa: odcinam dół i górę, gdzie jest tylko szum
PASMO = (100.0, 1500.0) 
# Liczba cech MFCC - 20, jest to standard w analizie mowy i dźwięków medycznych
CECHY_COUNT = 20

# Funkcja projektująca filtr Butterwortha (rząd 4 zapewnia strome zbocze odcięcia)
def filtr_pasmo(data, low, high, fs, order=4):
    nyq = 0.5 * fs # Częstotliwość Nyquista (połowa próbkowania)
    # Normalizacja częstotliwości względem Nyquista dla funkcji lfilter
    b, a = butter(order, [low/nyq, high/nyq], btype='band')
    return lfilter(b, a, data)

# Tworzę słownik ścieżek, żeby nie martwić się błędami w nazwach folderów w CSV (Był z tym wcześniej problem)
pliki_na_dysku = {}
for root, dirs, files in os.walk(BASE_PATH):
    for f in files:
        if f.lower().endswith('.wav'):
            # Kluczem jest sama nazwa pliku (małe litery), wartością pełna ścieżka C:\...
            pliki_na_dysku[f.strip().lower()] = os.path.join(root, f)

def wyciagnij_mfcc(sciezka, label):
    try:
        # Wczytuję plik z resamplem do 4000 Hz (starczy dla pasma 100-1500Hz)
        sygnal, fs = librosa.load(sciezka, sr=4000)
        # Nakładam filtr pasmowy, żeby pozbyć się np. stuknięć w mikrofon
        przefiltrowany = filtr_pasmo(sygnal, PASMO[0], PASMO[1], fs)
        
        # Obliczanie współczynników MFCC (cechy psychoakustyczne)
        mfccs = librosa.feature.mfcc(y=przefiltrowany, sr=fs, n_mfcc=CECHY_COUNT)
        
        # Agreguję dane czasowe do statystyk (średnia i odchylenie dla każdego MFCC)
        wynik = {'label': label}
        for i in range(CECHY_COUNT):
            # Średnia mówi o ogólnym brzmieniu danej częstotliwości
            wynik[f'mfcc_mean_{i}'] = np.mean(mfccs[i])
            # Odchylenie standardowe mówi o dynamice (zmianach) dźwięku
            wynik[f'mfcc_std_{i}'] = np.std(mfccs[i])
        return wynik
    except Exception as e:
        return None # Jeśli plik jest uszkodzony, pomijam go

# Pętla główna: tworzy features_train.csv, features_val.csv itd.
for plik_csv in PLIKI_CSV:
    df = pd.read_csv(os.path.join(BASE_PATH, plik_csv))
    lista_cech = []
    
    print(f"Przetwarzam pliki z listy: {plik_csv}...")
    for _, row in df.iterrows():
        # Wyciągam nazwę pliku z kolumny 'filename' i czyszczę z białych znaków
        nazwa = os.path.basename(row['filename']).strip().lower()
        if nazwa in pliki_na_dysku:
            dane = wyciagnij_mfcc(pliki_na_dysku[nazwa], row['label'])
            if dane:
                lista_cech.append(dane)
    
    # Zapisuję wyniki do tabeli - to będzie bezpośredni wsad do modeli
    pd.DataFrame(lista_cech).to_csv(os.path.join(BASE_PATH, f"features_{plik_csv}"), index=False)

print("Eksperyment zakończony sukcesem. Cechy wyeksportowane.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
import os

FOLDER = 'C:/Users/macie/Desktop/PROJEKT/'

# Ładowanie danych przygotowanych przez poprzedni skrypt
train_df = pd.read_csv(os.path.join(FOLDER, 'features_train.csv'))
test_df = pd.read_csv(os.path.join(FOLDER, 'features_test.csv'))

# Rozdzielenie cech (X) od etykiet (y - Normal/Pathology)
X_train, y_train = train_df.drop('label', axis=1), train_df['label']
X_test, y_test = test_df.drop('label', axis=1), test_df['label']

# Skalowanie danych (Z-score normalization)
# SVM jest wrażliwy na skalę - MFCC mają różne rzędy wielkości, musimy je ujednolicić
skaler = StandardScaler()
X_train_s = skaler.fit_transform(X_train) # Uczy się średniej i odchylenia na treningu
X_test_s = skaler.transform(X_test)      # Aplikuje te same parametry na teście

# --- OPTYMALIZACJA HIPERPARAMETRÓW (GridSearch) ---
# Szukamy najlepszej kombinacji C (kara za błędy) i jądra (sposób mapowania danych)
print("Uruchamiam GridSearch w celu znalezienia optymalnego modelu")
param_grid = {
    'C': [0.1, 1, 10, 100],          # Parametr regularyzacji
    'gamma': ['scale', 'auto', 0.01], # Szerokość jądra RBF
    'kernel': ['rbf', 'poly']         # Rodzaj jądra (nieliniowe)
}

# 5-krotna walidacja krzyżowa (CV=5) gwarantuje, że wyniki nie są dziełem przypadku
grid = GridSearchCV(SVC(), param_grid, cv=5, verbose=0)
grid.fit(X_train_s, y_train)

best_model = grid.best_estimator_
print(f"Wybrane parametry: {grid.best_params_}")

# Predykcja na zbiorze testowym, którego model wcześniej "nie widział"
y_pred = best_model.predict(X_test_s)

# Wyświetlenie raportu (Precision, Recall, F1-Score)
print("\nStatystyki klasyfikacji:")
print(classification_report(y_test, y_pred))

# Tworzenie Macierzy Pomyłek - pokazuje ile razy model się pomylił (FP/FN)
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
# Użyjemy Seaborna do wizualizacji mapy ciepła (heatmap)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=best_model.classes_, 
            yticklabels=best_model.classes_)
plt.title('Macierz pomyłek - Model SVM (Maciej)')
plt.ylabel('Prawdziwa diagnoza')
plt.xlabel('Przewidywanie modelu')
# Zapisuję obraz
plt.savefig(os.path.join(FOLDER, 'macierz_pomyłek_SVM.png'))
plt.show()

# Automatyczny zapis raportu do pliku tekstowego (dla dokumentacji)
with open(os.path.join(FOLDER, 'wyniki_svm_dokumentacja.txt'), 'w') as f:
    f.write(f"Najlepsze parametry GridSearch: {grid.best_params_}\n\n")
    f.write(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
from scipy.signal import butter, lfilter
import os

BASE = 'C:/Users/macie/Desktop/PROJEKT/'
CSV_LIST = ['train.csv', 'val.csv', 'test.csv']

# Ponownie używam filtra Butterwortha rzędu 4 dla spójności danych
def dsp_filter(signal, fs):
    nyq = 0.5 * fs
    b, a = butter(4, [100/nyq, 1500/nyq], btype='band')
    return lfilter(b, a, signal)

# Szybkie skanowanie dysku w poszukiwaniu audio
audio_index = {}
for root, _, files in os.walk(BASE):
    for f in files:
        if f.lower().endswith('.wav'):
            audio_index[f.strip().lower()] = os.path.join(root, f)

for plik_csv in CSV_LIST:
    data = pd.read_csv(os.path.join(BASE, plik_csv))
    folder_name = plik_csv.split('.')[0] # train, val lub test
    
    print(f"Generuję spektrogramy dla zbioru: {folder_name}")
    
    for _, row in data.iterrows():
        clean_name = os.path.basename(row['filename']).strip().lower()
        
        if clean_name in audio_index:
            # Tworzę podfoldery zgodne z wymogami bibliotek do CNN
            save_dir = os.path.join(BASE, 'DANE_OBRAZOWE_CNN', folder_name, row['label'])
            os.makedirs(save_dir, exist_ok=True)
            
            save_path = os.path.join(save_dir, clean_name.replace('.wav', '.png'))
            
            # Jeśli obrazek już jest, nie tracę czasu na ponowne liczenie
            if not os.path.exists(save_path):
                try:
                    audio, sr = librosa.load(audio_index[clean_name], sr=4000)
                    filtered = dsp_filter(audio, sr)
                    
                    # Rysowanie spektrogramu (skala decybelowa)
                    plt.figure(figsize=(4, 4))
                    # STFT - Krótkookresowa Transformacja Fouriera
                    stft_data = librosa.stft(filtered)
                    # Konwersja amplitudy na dB (logarytmiczna skala głośności)
                    db_data = librosa.amplitude_to_db(np.abs(stft_data))
                    
                    # Wyświetlenie spektrogramu bez osi i napisów - czysty wsad do CNN
                    librosa.display.specshow(db_data, sr=sr)
                    plt.axis('off') # Ukrycie osi czasu i częstotliwości
                    plt.savefig(save_path, bbox_inches='tight', pad_inches=0)
                    plt.close() # Zamknięcie wykresu z pamięci RAM
                except:
                    continue
```
