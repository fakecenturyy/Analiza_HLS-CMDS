```python
import pandas as pd
import numpy as np
import librosa
import os
from scipy.signal import butter, lfilter

# --- KONFIGURACJA ---
BASE_PATH = 'C:/Users/macie/Desktop/PROJEKT/'
CSV_FILES = ['train.csv', 'val.csv', 'test.csv']
F_LOW, F_HIGH = 100.0, 1500.0 
N_MFCC = 20

def butter_bandpass_filter(data, lowcut, highcut, fs, order=4):
    nyq = 0.5 * fs
    low, high = lowcut / nyq, highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return lfilter(b, a, data)

# Mapowanie plików (skanowanie dysku)
file_map = {}
for root, dirs, files in os.walk(BASE_PATH):
    for file in files:
        if file.lower().endswith('.wav'):
            file_map[file.strip().lower()] = os.path.join(root, file)

def extract_row_features(file_path, label):
    try:
        y, fs = librosa.load(file_path, sr=4000)
        y_filt = butter_bandpass_filter(y, F_LOW, F_HIGH, fs)
        mfccs = librosa.feature.mfcc(y=y_filt, sr=fs, n_mfcc=N_MFCC)
        
        row = {'label': label}
        for i in range(N_MFCC):
            row[f'mfcc_mean_{i}'] = np.mean(mfccs[i])
            row[f'mfcc_std_{i}'] = np.std(mfccs[i])
        return row
    except:
        return None

# --- PROCESOWANIE ---
for csv_name in CSV_FILES:
    df = pd.read_csv(os.path.join(BASE_PATH, csv_name))
    output_name = f"features_{csv_name}"
    print(f"Generowanie cech dla {csv_name}...")
    
    features_list = []
    for _, row in df.iterrows():
        fname = os.path.basename(row['filename']).strip().lower()
        if fname in file_map:
            feat = extract_row_features(file_map[fname], row['label'])
            if feat: features_list.append(feat)
            
    pd.DataFrame(features_list).to_csv(os.path.join(BASE_PATH, output_name), index=False)

print("✅ Gotowe! Masz komplet plików features_*.csv")
```

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler
import os

BASE_PATH = 'C:/Users/macie/Desktop/PROJEKT/'

# 1. Wczytanie danych
train = pd.read_csv(os.path.join(BASE_PATH, 'features_train.csv'))
test = pd.read_csv(os.path.join(BASE_PATH, 'features_test.csv'))

# 2. Podział na X i y
X_train = train.drop('label', axis=1)
y_train = train['label']
X_test = test.drop('label', axis=1)
y_test = test['label']

# 3. Skalowanie (BARDZO WAŻNE DLA SVM)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Trening modelu SVM
print("Trenuję model SVM (Maciek)...")
model = SVC(kernel='rbf', C=10, gamma='scale') # C=10 zazwyczaj lepiej działa na MFCC
model.fit(X_train_scaled, y_train)

# 5. Wyniki
y_pred = model.predict(X_test_scaled)
print("\n=== WYNIKI MODELU SVM (MACIEK) ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%")
print("\nRaport:")
print(classification_report(y_test, y_pred))

# Zapisz wyniki do pliku dla Karola (on zrobi zestawienie)
with open(os.path.join(BASE_PATH, 'wyniki_svm_maciek.txt'), 'w') as f:
    f.write(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
from scipy.signal import butter, lfilter
import os

# --- KONFIGURACJA ---
BASE_PATH = 'C:/Users/macie/Desktop/PROJEKT/'
CSV_FILES = ['train.csv', 'val.csv', 'test.csv']
F_LOW, F_HIGH = 100.0, 1500.0 

def butter_bandpass_filter(data, lowcut, highcut, fs, order=4):
    nyq = 0.5 * fs
    low, high = lowcut / nyq, highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return lfilter(b, a, data)

def save_spectrogram(y, fs, save_path):
    plt.figure(figsize=(5, 5))
    S = librosa.stft(y)
    S_dB = librosa.amplitude_to_db(np.abs(S), ref=np.max)
    librosa.display.specshow(S_dB, sr=fs, cmap='viridis')
    plt.axis('off')
    plt.savefig(save_path, bbox_inches='tight', pad_inches=0)
    plt.close()

# 1. SKANOWANIE DYSKU (Budujemy mapę, gdzie są pliki .wav)
print("🔍 Skanuję folder PROJEKT w poszukiwaniu plików audio...")
file_map = {}
for root, dirs, files in os.walk(BASE_PATH):
    for file in files:
        if file.lower().endswith('.wav'):
            # Zapamiętujemy pełną ścieżkę do każdej nazwy pliku
            file_map[file.strip().lower()] = os.path.join(root, file)

print(f"✅ Znaleziono {len(file_map)} plików audio na dysku.")

# 2. GENEROWANIE WSZYSTKICH SPEKTROGRAMÓW
for csv_name in CSV_FILES:
    csv_path = os.path.join(BASE_PATH, csv_name)
    if not os.path.exists(csv_path):
        continue

    df = pd.read_csv(csv_path)
    split_type = csv_name.replace('.csv', '')
    print(f"\n--- Przetwarzanie zbioru: {split_type.upper()} ({len(df)} plików) ---")
    
    count = 0
    for _, row in df.iterrows():
        # Wyciągamy samą nazwę pliku z CSV, ignorując błędy w ścieżkach
        fname_in_csv = os.path.basename(row['filename']).strip().lower()
        
        if fname_in_csv in file_map:
            actual_file_path = file_map[fname_in_csv]
            
            # Budujemy ścieżkę zapisu: SPEKTROGRAMY_CNN / train / Normal / nazwa.png
            output_dir = os.path.join(BASE_PATH, 'SPEKTROGRAMY_CNN', split_type, row['label'])
            os.makedirs(output_dir, exist_ok=True)
            
            save_path = os.path.join(output_dir, fname_in_csv.replace('.wav', '.png'))
            
            # Generujemy, jeśli plik jeszcze nie istnieje
            if not os.path.exists(save_path):
                try:
                    y, fs = librosa.load(actual_file_path, sr=4000)
                    y_filt = butter_bandpass_filter(y, F_LOW, F_HIGH, fs)
                    save_spectrogram(y_filt, fs, save_path)
                    count += 1
                except Exception as e:
                    print(f"Błąd przy {fname_in_csv}: {e}")
        else:
            # Tutaj wypisze tylko te pliki, których naprawdę nie ma nigdzie w folderze PROJEKT
            pass 

    print(f"✅ Wygenerowano {count} nowych spektrogramów dla {split_type}.")

print("\n🚀 ZAKOŃCZONO! Cały folder 'SPEKTROGRAMY_CNN' jest gotowy dla Karola.")
```
